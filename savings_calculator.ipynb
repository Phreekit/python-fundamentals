```python
# ======================================================
# Day 1 Project: Income Target & Savings Calculator
# ======================================================

# 1. Gather User Inputs
name = input("Enter your name: ")
target_monthly_usd = float(input("Enter your target monthly income in USD ($): "))
savings_percent = float(input("Enter percentage of income to save (%): "))

# 2. Currency Conversion & Savings Calculations
usd_to_kes_rate = 129.0  # Exchange rate: $1 USD = 129 KES

monthly_income_kes = target_monthly_usd * usd_to_kes_rate
annual_income_kes = monthly_income_kes * 12

monthly_savings_kes = monthly_income_kes * (savings_percent / 100)
annual_savings_kes = monthly_savings_kes * 12

# 3. Print Formatted Results
print("\n" + "=" * 45)
print(f" 📊 FINANCIAL TARGETS FOR {name.upper()}")
print("=" * 45)
print(f"• Monthly Income Target (USD): ${target_monthly_usd:,.2f}")
print(f"• Monthly Income Target (KES): KSh {monthly_income_kes:,.2f}")
print(f"• Annual Income Target (KES):  KSh {annual_income_kes:,.2f}")
print("-" * 45)
print(f"• Monthly Savings Goal ({savings_percent:.0f}%):  KSh {monthly_savings_kes:,.2f}")
print(f"• Annual Savings Goal ({savings_percent:.0f}%):   KSh {annual_savings_kes:,.2f}")
print("=" * 45)
```
